“关闭”特性使 ReLU 成为一种非线性激活函数。为什么需要这一点？我们将在下面探讨。 

### 为什么需要非线性激活函数？  
<img align="left" src="./images/C2_W2_ReLU_Graph.png"     style=" width:250px; padding: 10px 20px; " > 图中所示的函数由若干线性片段组成（分段线性）。在线性部分，斜率保持不变，然后在过渡点处突然改变。在过渡点处，会加入一个新的线性函数，它与已有函数相加后会产生新的斜率。新函数在过渡点处加入，但在该点之前不影响输出。非线性激活函数负责在过渡点之前（有时也在其后）禁用输入。下面的练习提供了一个更具体的例子。

 
本练习的目标是体会 ReLU 的非线性行为如何提供了必要的能力：在需要之前将函数“关闭”。让我们看看在这个例子中是如何实现的。
<img align="right" src="./images/C2_W2_ReLU_Plot.png"     style=" width:600px; padding: 10px 20px; "> 
右侧的图包含第一层各单元的输出。   
从最上面开始，单元 0 负责标记为 1 的第一段。图中同时显示了线性函数 $z$ 和经过 ReLU 之后的函数 $a$。可以看到，ReLU 在区间 [0,1] 之后截断了该函数。这一点很重要，因为它避免了单元 0 干扰后续的段落。 

单元 1 负责第 2 段。在这里，ReLU 使该单元保持静默，直到 x 超过 1。由于第一个单元不再起作用，单元 1 的斜率 $w^{[1]}_1$ 就是目标直线的斜率。必须调整偏置，使输出在 x 达到 1 之前保持为负。注意单元 1 的作用也延伸到了第 3 段。

单元 2 负责第 3 段。ReLU 同样使输出保持为零，直到 x 达到合适的值。单元的斜率 $w^{[1]}_2$ 必须设置得当，使单元 1 和单元 2 之和具有所需的斜率。偏置同样需要调整，使输出在 x 达到 2 之前保持为负。 

ReLU 激活函数的“关闭”或禁用特性使模型能够把线性片段拼接起来，从而对复杂的非线性函数进行建模。
